```python
import anatomist.api as ana
from soma.qt_gui.qtThread import QtThreadCall
from soma.qt_gui.qt_backend import Qt

a = ana.Anatomist()

from soma import aims
from scipy import ndimage
import numpy as np
import glob
```

```python
side = 'L'
w = a.createWindow("3D")
dic_windows = {}
```

```python
density_for_masks = glob.glob(f"/neurospin/dico/data/deep_folding/current/mask/2mm/regions/{side}/*.nii.gz")
dic_bary = {}
for density in density_for_masks:
    vol = aims.read(density)
    region = density.replace(f"/neurospin/dico/data/deep_folding/current/mask/2mm/regions/{side}/", "")
    region = region.replace(".nii.gz", "")
    print(region)
    print("Volume under study:",(vol.np).shape, "\n")
    vol = vol.np[..., 0]
    x, y, z = np.indices(vol.shape)
    total_mass = vol.sum()
    print(total_mass)
    x_c = (x * vol).sum() / total_mass
    y_c = (y * vol).sum() / total_mass
    z_c = (z * vol).sum() / total_mass
    barycentre = (x_c, y_c, z_c)
    print(barycentre)
    dic_bary[region] = np.array(barycentre)
    barycentre_x, barycentre_y, barycentre_z = [int(i) for i in barycentre]
    barycentre_aims = np.zeros(vol.shape, dtype=np.int16)
    barycentre_aims[barycentre_x, barycentre_y, barycentre_z] = int(1)
    barycentre_aims = aims.Volume(barycentre_aims)
    barycentre_aims.header()['voxel_size'] = [2.0, 2.0, 2.0]
    file_nii = f"/volatile/ad279118/Figures_report/barycentre/bary_{region}.nii.gz"
    aims.write(barycentre_aims, file_nii)
```

```python
# Classification temporale vers intérieure

sorted_regions = sorted(dic_bary.items(), key=lambda kv: (kv[1][0], kv[1][1], kv[1][2]))

for region, coords in sorted_regions:
    print(region, coords)
```

```python
# Classification avant vers arrière

sorted_regions = sorted(dic_bary.items(), key=lambda kv: (kv[1][1], kv[1][2], kv[1][0]))

for region, coords in sorted_regions:
    print(region, coords)
```

```python
# Classification haut vers bas

sorted_regions = sorted(dic_bary.items(), key=lambda kv: (kv[1][2], kv[1][1], kv[1][0]))

for region, coords in sorted_regions:
    print(region, coords)
```
